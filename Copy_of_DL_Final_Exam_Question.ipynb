## Write Answer 02:Complete Backpropagation for a Neural Network with One Hidden Layer (Short Version)
1. Forward Pass
forward pass and backward pass. In the forward pass, input data is passed through the hidden layer and output layer using activation functions (such as sigmoid) to generate a prediction. The network's prediction is then compared with the actual target value, and a loss function is used to calculate the error.
In the backward pass, this error is propagated from the output layer back to the hidden layer using the chain rule of calculus. The contribution of each neuron and weight to the final error is computed by calculating gradients. These gradients indicate how much each weight and bias should change to reduce the error.

The weights and biases are then updated using gradient descent:

New Parameter=Old Parameter−η×Gradient\text{New Parameter} = \text{Old Parameter} - \eta \times \text{Gradient}New Parameter=Old Parameter−η×Gradient

where η\etaη is the learning rate.

The process of forward propagation, error calculation, backpropagation, and parameter updates is repeated for many epochs until the network minimizes the loss and learns the desired input-output relationship effectively.


# Question 03: [Marks 5]

## Discuss the vanishing gradient problems in deep neural networks. Explain why they occur during backpropagation, how activation functions contribute to this issue, and suggest at least three techniques to overcome these problems.


## Write Answer 03:
The vanishing gradient problem occurs in deep neural networks when the gradients become extremely small during backpropagation.
Techniques to Overcome the Problem:

Use ReLU Activation Function: ReLU has a derivative of 1 for positive inputs, helping gradients flow more effectively.
Proper Weight Initialization: Methods such as Xavier and He initialization help maintain gradient magnitude during training.
Batch Normalization: Normalizes inputs of each layer, improving gradient flow and training stability.
Residual Connections (ResNet): Skip connections allow gradients to pass directly to earlier layers, reducing gradient loss.
Use Advanced Optimizers: Adam and RMSProp can improve convergence and learning efficiency.

## Design and implement a deep Convolutional Neural Network (CNN) for a multi-class image classification problem using PyTorch. The network should consist of four convolutional blocks, where each block includes two convolution layers followed by activation and normalization, and a pooling layer at the end. The number of filters should increase progressively across blocks. After the convolutional part, flatten the feature maps and connect them to fully connected layers with dropout for regularization, and produce final outputs corresponding to K classes. Also explain the role of each part of the network in detail.

# Write Answer 10:
Role of Each Part
Convolution Layers: Extract image features such as edges, textures, and shapes.
ReLU Activation: Introduces non-linearity and speeds up training.
Batch Normalization: Stabilizes learning and improves convergence.
Max Pooling: Reduces feature-map size and computational cost.
Increasing Filters (32→64→128→256): Learns progressively more complex features.
Flatten Layer: Converts feature maps into a 1D vector.
Fully Connected Layers: Perform final classification.
Dropout: Prevents overfitting by randomly disabling neurons during training.
Output Layer (K classes): Produces class scores for multi-class classification.

```python
import torch
import torch.nn as nn
import torch.optim as optim

class DeepCNN(nn.Module):
    def __init__(self, K):
        super(DeepCNN, self).__init__()

        self.features = nn.Sequential(

            # Block 1
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.BatchNorm2d(32),
            nn.Conv2d(32, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.BatchNorm2d(32),
            nn.MaxPool2d(2),

            # Block 2
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.BatchNorm2d(64),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.BatchNorm2d(64),
            nn.MaxPool2d(2),

            # Block 3
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.BatchNorm2d(128),
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.BatchNorm2d(128),
            nn.MaxPool2d(2),

            # Block 4
            nn.Conv2d(128, 256, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.BatchNorm2d(256),
            nn.Conv2d(256, 256, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.BatchNorm2d(256),
            nn.MaxPool2d(2)
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256 * 4 * 4, 512),  # for 64x64 input images
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(512, K)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x


# Example
K = 10  # Number of classes
model = DeepCNN(K)

print(model)
```
